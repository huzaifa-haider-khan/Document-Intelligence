```python
# Cell 4 - Task 1.1: Load and Prepare Data
def load_documents(data_dir):
    """Load documents from folder structure"""
    documents = []
    labels = []

    for doc_type in os.listdir(data_dir):
        folder_path = os.path.join(data_dir, doc_type)
        if not os.path.isdir(folder_path):
            continue
        for filename in os.listdir(folder_path):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, 'r') as f:
                text = f.read()
            documents.append(text)
            labels.append(doc_type)

    return documents, labels

# Load data
documents, labels = load_documents('training_data')

print(f'Total documents loaded : {len(documents)}')
print(f'Classes found          : {set(labels)}')
print(f'Samples per class      : {len(documents)//len(set(labels))}')
print('\n✅ Data loading complete!')
```

```python
# Cell 7 - Task 2.1: Create FastAPI App
api_code = '''
from fastapi import FastAPI, UploadFile, File
from fastapi.responses import JSONResponse
import joblib
import io

# Initialize FastAPI
app = FastAPI(
    title="Document Intelligence API",
    description="OCR, Classification, and Extraction",
    version="1.0.0"
)

# Load models
vectorizer = joblib.load('../models/vectorizer.pkl')
classifier = joblib.load('../models/classifier.pkl')

@app.get("/")
def root():
    return {
        "message": "Document Intelligence API",
        "version": "1.0.0",
        "endpoints": ["/classify", "/extract", "/process"]
    }
'''

# Save to api/main.py
with open('api/main.py', 'w') as f:
    f.write(api_code)

print('✅ api/main.py created successfully!')

# Verify file
with open('api/main.py', 'r') as f:
    print('\nFile contents:')
    print(f.read())
```

```python
# Cell 8 - Task 2.2: Add Classification Endpoint
classify_code = '''
@app.post("/classify")
async def classify_document(file: UploadFile = File(...)):
    """Classify document type"""
    try:
        # Read file contents
        contents = await file.read()
        text = contents.decode("utf-8")

        # Vectorize and classify
        text_vec = vectorizer.transform([text])
        prediction = classifier.predict(text_vec)[0]
        probabilities = classifier.predict_proba(text_vec)[0]
        confidence = max(probabilities)

        return {
            "document_type": prediction,
            "confidence": float(confidence),
            "all_probabilities": {
                cls: float(prob)
                for cls, prob in zip(classifier.classes_, probabilities)
            }
        }
    except Exception as e:
        return JSONResponse(status_code=500, content={"error": str(e)})
'''

# Append to api/main.py
with open('api/main.py', 'a') as f:
    f.write(classify_code)

print('✅ /classify endpoint added to api/main.py!')
```

```python
# Cell 9 - Task 2.3: Add Extraction Endpoint
extract_code = '''
import re

def extract_dates(text):
    patterns = [
        r"\\d{1,2}/\\d{1,2}/\\d{4}",
        r"\\d{1,2}-\\d{1,2}-\\d{4}",
        r"\\b(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)[a-z]* \\d{1,2},? \\d{4}",
        r"\\d{4}-\\d{2}-\\d{2}"
    ]
    dates = []
    for pattern in patterns:
        dates.extend(re.findall(pattern, text, re.IGNORECASE))
    return dates

def extract_amounts(text):
    pattern = r"\\$?\\d{1,3}(?:,\\d{3})*(?:\\.\\d{2})?"
    amounts = re.findall(pattern, text)
    cleaned = []
    for amount in amounts:
        clean = amount.replace("$", "").replace(",", "")
        try:
            cleaned.append(float(clean))
        except ValueError:
            pass
    return cleaned

def extract_invoice_number(text):
    patterns = [
        r"INV-\\d{4}-\\d{3}",
        r"#\\d{5,}",
        r"ORDER-[A-Z0-9]+",
        r"Invoice (?:Number|#):?\\s*([A-Z0-9-]+)"
    ]
    for pattern in patterns:
        match = re.search(pattern, text, re.IGNORECASE)
        if match:
            return match.group(1) if match.groups() else match.group(0)
    return None

@app.post("/extract")
async def extract_information(file: UploadFile = File(...)):
    """Extract information from document"""
    try:
        contents = await file.read()
        text = contents.decode("utf-8")

        dates = extract_dates(text)
        amounts = extract_amounts(text)
        invoice_number = extract_invoice_number(text)

        return {
            "dates":          dates,
            "amounts":        amounts,
            "invoice_number": invoice_number,
            "raw_text":       text[:500]
        }
    except Exception as e:
        return JSONResponse(status_code=500, content={"error": str(e)})
'''

# Append to api/main.py
with open('api/main.py', 'a') as f:
    f.write(extract_code)

print('✅ /extract endpoint added to api/main.py!')
```

```python
# Cell 10 - Task 2.4: Add Complete Processing Endpoint
process_code = '''
@app.post("/process")
async def process_document(file: UploadFile = File(...)):
    """Complete pipeline: Classify + Extract"""
    try:
        contents = await file.read()
        text = contents.decode("utf-8")

        # Classify
        text_vec = vectorizer.transform([text])
        doc_type = classifier.predict(text_vec)[0]
        confidence = max(classifier.predict_proba(text_vec)[0])

        # Extract
        extracted_data = {
            "dates":          extract_dates(text),
            "amounts":        extract_amounts(text),
            "invoice_number": extract_invoice_number(text)
        }

        return {
            "document_type":  doc_type,
            "confidence":     float(confidence),
            "extracted_data": extracted_data,
            "status":         "success"
        }
    except Exception as e:
        return JSONResponse(
            status_code=500,
            content={"error": str(e), "status": "failed"}
        )
'''

# Append to api/main.py
with open('api/main.py', 'a') as f:
    f.write(process_code)

print('✅ /process endpoint added to api/main.py!')
```
